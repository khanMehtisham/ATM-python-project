```python
# Write a function to check if a given string is a palindrome. The function should take a string 
# as input and return True if it's a palindrome, otherwise False.

#(string is said to be palindrome if it is same in reverse order also e.g: civic, radar etc)





def palindrome(s):
    i=0
    n=-1
    y=0
    m=0
    c=len(s)
    #print(1)
    #print(c)
    m=c/2
    while y<m:
        if s[i]==s[n]:
            n-=1
            i+=1
            
        else:
            #print("no it is not")
            return('false')
            break
        y+=1
    if s[i]==s[n]:
        #print("yes it is")
        return('true')
        
                
                
                        
    
s=str(input("enter any string"))
z=palindrome(s)
print(z)    
    
    
    

```

```python
# Write a function that takes a list of integers and returns two separate lists:
#one with even numbers and the other with odd numbers.
def numbers(l):
    l1=[]
    l2=[]
    for x in l:
        if x%2==0:
            l1.append(x)
        else:
            l2.append(x)

    #print(l1,l2)
    return(l1,l2)
            



l=[2,5,8,7,9,4,6]
even,odd=numbers(l)
print("even list",even)
print("odd list",odd)
```
