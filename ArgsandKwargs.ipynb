```python
def odd_even(*args):
    for x in args:
        if x%2==0:
            print("{} is even number".format(x))
        else:
            print("{} is odd number".format(x))
            


odd_even(2,6,7,8,3,9,1,27,48)
```

```python
def info(**kwargs):
    for(x,y) in kwargs.items():
        print(f"{x}:{y}")


info(name="saba",c=8,section="green",rollno=101,age=21,address="srinagar",UID=5467895646,contact=7898765678)
```

```python
def join(*args,**kwargs):
    for x in args:
        if x%2==0:
            print("{} is even number".format(x))
        else:
            print("{} is odd number".format(x))

    for (a,b) in kwargs.items():
         print(f"{a}:{b}")
        
        


join(45,67,9,42,26,name="saba",c=8,section="green",rollno=101,age=21,address="soura")
```
