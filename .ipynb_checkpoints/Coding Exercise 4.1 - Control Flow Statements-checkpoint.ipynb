```python
# Create a program that checks whether a number is divisible by both 3 and 5. 
# If yes, print "Divisible by both", otherwise, print "Not divisible by both".
num=10
if num%3==0 and num%5==0:
    print("num is divisible by both 3 and 5")
else:
    print("Not divisible by both")
    
```

```python
# Given a student's score, print their grade using the following criteria:

# 90 and above: "A"
# 80-89: "B"
# 70-79: "C"
# 60-69: "D"
# Below 60: "F"
marks=55
if marks>=90:
    print("Grade A")
elif marks>=80 and marks<=89:
    print("Grade B")
elif marks>=70 and marks<=79:
    print("Grade C")
elif marks>=60 and marks<=69:
    print("Grade D")
else:
    print("Grade F")
                
                
```

```python
# Take the lengths of three sides of a triangle and determine its type:

# Equilateral (all sides are equal)
# Isosceles (two sides are equal)
# Scalene (all sides are different)
s1=36
s2=16
s3=36
if s1==s2 and s2==s3 and s1==s3: 
    print("Equilateral")
elif s1==s2 or s2==s3 or s1==s3:
     print("Isosceles")
else:
    print("Scalene")

    
```

```python
# Check if a number lies within a specific range and categorize it:

# 0-50: "Low"
# 51-100: "Medium"
# Above 100: "High"
num=85
if num>=0 and num<=50:
    print("low")
elif num>=51 and num<=100:
    print("Medium")
elif num>100:
    print("High")
            
```

```python
# Given the coordinates (x, y), determine which quadrant of the Cartesian plane the point lies in:

# Quadrant 1: x > 0 and y > 0
# Quadrant 2: x < 0 and y > 0
# Quadrant 3: x < 0 and y < 0
# Quadrant 4: x > 0 and y < 0
x=-1
y=-3
if x > 0 and y > 0:
    print("Quadrant 1")
elif x < 0 and y > 0:
    print("Quadrant 2")
elif x < 0 and y < 0:
    print("Quadrant 3")
elif x < 0 and y < 0:
    print("Quadrant 4")
else:
    print("NA")
```

```python
# Given a year, determine whether it falls in the 20th century (1901-2000), 
# the 21st century (2001-2100), or earlier.
year=2001
if year>=1901 and year<=2000:
    print("it falls in the 20th century")
elif year>=2001 and year<=2100:
    print("it falls in the 21st century")
else:
    print("other century")
```

```python
# Write a program that categorizes a person’s age into the following groups:

# 0-12: Child
# 13-19: Teenager
# 20-59: Adult
# 60 and above: Senior Citizen
age=19
if age>=0 and age<=12:
    print("child")
elif age>=13 and age<=19:
    print("Teenager") 
elif age>=20 and age<=59:
    print("Adult")
elif age>=60:
    print("Senior Citizen")
```

```python
# Classify weather based on temperature:

# <= 0: "Freezing"
# 1-10: "Cold"
# 11-25: "Moderate"
# > 25: "Hot"
temp=49
if temp<=0:
    print("Freezing")
elif temp>=1 and temp<=10:
     print("Cold")
elif temp>=11 and temp<=25:
    print("Moderate")
elif temp>25:
    print("Hot")
```

```python
# Given a character, check if it is:

# Uppercase
# Lowercase
# Not a letter
x='4'
if x.isupper():
    print("Uppercase")
elif x.islower():
    print("Lowercase")
else:
    print("Not a letter")
```

```python
# Write a program that takes a number (1-7) representing a day of the week and prints its name (e.g., 1 = Monday, 7 = Sunday).
day=4
if day==1:
    print("Monday")
elif day==2:
     print("Tuesday")
elif day==3:
     print("Wednesday")
elif day==4:
     print("Thursday")
elif day==5:
     print("Friday")
elif day==6:
     print("Saturday")
elif day==7:
     print("Sunday")
else:
    print("Invalid number")
```
