```python
# Write a function check_even_odd(num) that checks whether a given number is even or odd.

def check_even_odd(num):
    if num%2==0:
        print("even number")
    else:
        print("odd number")




#num=10
num=int(input("enter a positive number"))
check_even_odd(num)
```

```python
# Write a function count_vowels(s) that takes a string and returns the count of vowels (a, e, i, o, u) in it.


def count_vowels(s):
    s1="aeiou"
    count=0
    for x in s:
        if x in s1:
            count+=1
        else:
            continue

    print(count)
    



s="machine learning"
count_vowels(s)
```

```python
# Write a function remove_duplicates(lst) that takes a list and returns a new list with duplicates removed (use sets).

def remove_duplicates(lst):
    l1=set(lst)
    return(l1)



lst=[2,5,7,4,3,8,2,5,4]
result=remove_duplicates(lst)
print(result)
```

```python
# Write a fun1ction key_exists(d, key) that checks if a given key exists in a dictionary.
def key_exists(d, key):
    if key in d:
        print("key exists")
    else:
        print("key does not exist")
            



d={'k1':10,'k2':20,'k3':30}
key=str(input("enter key"))
key_exists(d, key)

    
    
```

```python
# Write a function reverse_string(s) that returns the reverse of a given string.
def r_string(s):
    return(s[::-1])




#s="hello"
s=str(input("enter any string"))
result=r_string(s)
print(result)
```

```python
# Write a function set_intersection(set1, set2) that returns the intersection of two sets.

set1={1,2,3,4,6,7,9}
set2={3,4,7,9,1,5,0}

#intersection is common or matching elements between set1 and set2 (you have to print those only)

def intersection(set1,set2):
    set3=set()
    for x in set1:
        for y in set2:
            if x==y:
                set3.add(x)
            else:
                continue

    print(set3)
                
set1={1,2,3,4,6,7,9}
set2={3,4,7,9,1,5,0}
intersection(set1,set2)
```
