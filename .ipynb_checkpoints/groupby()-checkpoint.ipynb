```python
# Q1)

# Given a DataFrame with columns Department, Employee, and Salary, group the
# data by Department and calculate the average salary for each department.
                                                                                                                              
# import pandas as pd

# data = {
#     'Department': ['HR', 'HR', 'IT', 'IT', 'Finance', 'Finance'],
#     'Employee': ['Alice', 'Bob', 'Charlie', 'David', 'Eva', 'Frank'],
#     'Salary': [50000, 55000, 60000, 65000, 70000, 72000]
# }
# df = pd.DataFrame(data)

# # Your code here
import pandas as pd
data = {
     'Department': ['HR', 'HR', 'IT', 'IT', 'Finance', 'Finance'],
     'Employee': ['Alice', 'Bob', 'Charlie', 'David', 'Eva', 'Frank'],
     'Salary': [50000, 55000, 60000, 65000, 70000, 72000]
 }
df = pd.DataFrame(data)

print(df.groupby("Department")["Salary"].mean())
```

```python
# Q2)

# Using the same DataFrame above, calculate:

# Total salary per department

# Number of employees per department
import pandas as pd
data = {
     'Department': ['HR', 'HR', 'IT', 'IT', 'Finance', 'Finance'],
     'Employee': ['Alice', 'Bob', 'Charlie', 'David', 'Eva', 'Frank'],
     'Salary': [50000, 55000, 60000, 65000, 70000, 72000]
 }
df = pd.DataFrame(data)
#print(df.groupby("Department")['Salary'].agg(['sum']))
print(df.groupby("Department")['Salary'].sum())
print(df.groupby("Department")['Employee'].count())


```

```python
# Q3)

# Using the same data, group by Department and find out which department has the highest average salary.

import pandas as pd
data = {
     'Department': ['HR', 'HR', 'IT', 'IT', 'Finance', 'Finance'],
     'Employee': ['Alice', 'Bob', 'Charlie', 'David', 'Eva', 'Frank'],
     'Salary': [50000, 55000, 60000, 65000, 70000, 72000]
 }
df = pd.DataFrame(data)
avg_sal=df.groupby("Department")["Salary"].mean()
print(avg_sal.idxmax())
print(max(avg_sal))



```

```python
# Group the DataFrame by Department and return the maximum and minimum salary in each department using .agg().


import pandas as pd
data = {
     'Department': ['HR', 'HR', 'IT', 'IT', 'Finance', 'Finance'],
     'Employee': ['Alice', 'Bob', 'Charlie', 'David', 'Eva', 'Frank'],
     'Salary': [50000, 55000, 60000, 65000, 70000, 72000]
 }
df = pd.DataFrame(data)
print(df.groupby("Department")["Salary"].agg(['max','min']))
```

```python
# Q4)

# Given the following DataFrame, calculate the average salary grouped by both Department and Gender.
# data = {
#     'Department': ['HR', 'HR', 'IT', 'IT', 'Finance', 'Finance'],
#     'Gender': ['F', 'M', 'M', 'M', 'F', 'M'],
#     'Salary': [50000, 55000, 60000, 65000, 70000, 72000]
# }
# df = pd.DataFrame(data)

# # Your code here
import pandas as pd
data = {
     'Department': ['HR', 'HR', 'IT', 'IT', 'Finance', 'Finance'],
     'Gender': ['F', 'M', 'M', 'M', 'F', 'M'],
     'Salary': [50000, 55000, 60000, 65000, 70000, 72000]
 }
df = pd.DataFrame(data)

print(df.groupby(["Department","Gender"])["Salary"].mean())


```

```python
# Q5)

# Using the DataFrame above, count the number of males and females in each department.

import pandas as pd
data = {
     'Department': ['HR', 'HR', 'IT', 'IT', 'Finance', 'Finance'],
     'Gender': ['F', 'M', 'M', 'M', 'F', 'M'],
     'Salary': [50000, 55000, 60000, 65000, 70000, 72000]
 }
df = pd.DataFrame(data)
print(df.groupby(["Department","Gender"]).count())
#print(df.groupby("Department")['Gender'].count())
```
