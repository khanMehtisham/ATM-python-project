```python
# Use this Dataset

import pandas as pd

data = {
    "Name": ["Ali", "Meesum", "Shifaan", "Fiza", "Kabir",
             "Danish", "Kumail", "Rukbaan", "Sharik", "Zoya"],
    "Department": ["IT", "CS", "IT", "CS", "ECE",
                   "ECE", "IT", "CS", "ECE", "IT"],
    "Marks": [78, 88, 65, 95, 72, 84, 90, 76, 69, 82],
    "Attendance": [88, 92, 75, 96, 80, 85, 91, 78, 72, 89]
}

df = pd.DataFrame(data)
#using idxmax function to find the department with maximum marks and maximum attendance
Total=df.groupby("Department")["Marks"].sum()
print(Total.idxmax())
print(max(Total))
att=df.groupby("Department")["Attendance"].sum()
print(att.idxmax())
print(max(att))
```

```python
# LAMBDA WITH PANDAS
# Use a lambda function with df["Marks"].apply()
# to create a new column called "Bonus_Marks".
#
# Rule:
# If marks are >= 80, give 5 bonus marks.
# Otherwise, give 2 bonus marks.
# Display the Name, Marks and Bonus_Marks columns.
# Write your code below.

import pandas as pd
data = {
    "Name": ["Ali", "Meesum", "Shifaan", "Fiza", "Kabir",
             "Danish", "Kumail", "Rukbaan", "Sharik", "Zoya"],
    "Department": ["IT", "CS", "IT", "CS", "ECE",
                   "ECE", "IT", "CS", "ECE", "IT"],
    "Marks": [78, 88, 65, 95, 72, 84, 90, 76, 69, 82],
    "Attendance": [88, 92, 75, 96, 80, 85, 91, 78, 72, 89]
}

df = pd.DataFrame(data)
#print(df["Name"])
df['Bonus_Marks']=df["Marks"].apply(lambda x:x+5 if x>=80 else x+2)

print(df[["Name","Marks","Bonus_Marks"]])
```

```python

# Create a function called calculate_average(*marks).
#
# It should:
# 1. Accept any number of marks.
# 2. Return their average.
#
# Test it with:
# calculate_average(70, 80, 90)
# calculate_average(65, 75, 85, 95)

# Write your code below.

def calculate_average(*args):
    x=sum(args)
    l=len(args)
    average=x/l
    print(average)


calculate_average(70, 80, 90)
calculate_average(65,75,85,95)
```

```python

# Create a function called student_details(**kwargs).
# The function should print all the information passed to it.
# Test it using:
# name, department, marks and attendance
# Example idea:
# student_details(name="Ali", department="IT", marks=78, attendance=88)
# Write your code below.

def student_details(**kwargs):
    for (x,y) in kwargs.items():
        print(f"{x}:{y}")


student_details(name="Ali", department="IT", marks=78, attendance=88)
```

```python
# Create a function called student_report(*marks, **details).
# The function should:
# 1. Calculate the average of the marks received through *marks.
# 2. Get the student's name and department from **details.
# 3. Return a dictionary containing:
#    name
#    department
#    average
# Test it with at least two students.
# Write your code below.
def student_report(*marks, **details):
    d=dict()
    x=sum(marks)
    l=len(marks)
    a=x/l
    #print(a)
    z="average"
    for x,y in details.items():
        d[x]=y
       
    d[z]=a    
    print(d)

  
student_report(40,80,70,90, name="saba",department="science")
student_report(78,87,96,98, name="amir",department="science")
```

```python

# Create a function called safe_divide(a, b).
#
# Divide a by b.
# Handle ZeroDivisionError if b is 0.
# Also handle TypeError if non-numeric values are supplied.
#
# Test:
# safe_divide(100, 5)
# safe_divide(100, 0)
# safe_divide("100", 5)

# Write your code below.

def safe_divide(a, b):
    d=0
    try:
        d=a/b
        print(d)
    except ZeroDivisionError:
        print("Dividing a number by zero")
    except TypeError:
        print("enter numeric values")
        


#x=int(input("enter first number"))
#y=int(input("enter second number"))
safe_divide(100,5)
safe_divide(100,0)
safe_divide("100",5)
```

```python
# XCEPTION HANDLING WITH DATA
# Create a function called get_mark(df, student_name).
# It should return the marks of the requested student.
# Handle the situation where the student does not exist
# by displaying a suitable message instead of crashing.
# Test it with:
# "Ali"
# "Unknown"
# Write your code below.

import pandas as pd
data = {
    "Name": ["Ali", "Meesum", "Shifaan", "Fiza", "Kabir",
             "Danish", "Kumail", "Rukbaan", "Sharik", "Zoya"],
    "Department": ["IT", "CS", "IT", "CS", "ECE",
                   "ECE", "IT", "CS", "ECE", "IT"],
    "Marks": [78, 88, 65, 95, 72, 84, 90, 76, 69, 82],
    "Attendance": [88, 92, 75, 96, 80, 85, 91, 78, 72, 89]
}

df = pd.DataFrame(data)

def get_mark(df,student):
    try:
        student=df[df["Name"]==student]
        marks=student["Marks"].values[0]
        return marks
    except IndexError: 
        print("you have entered the wrong name")
  

```

```python

# Use groupby("Department") to calculate BOTH:
# 1. Average Marks
# 2. Average Attendance
# Display the department-wise summary.
# Write your code below.

import pandas as pd
data = {
    "Name": ["Ali", "Meesum", "Shifaan", "Fiza", "Kabir",
             "Danish", "Kumail", "Rukbaan", "Sharik", "Zoya"],
    "Department": ["IT", "CS", "IT", "CS", "ECE",
                   "ECE", "IT", "CS", "ECE", "IT"],
    "Marks": [78, 88, 65, 95, 72, 84, 90, 76, 69, 82],
    "Attendance": [88, 92, 75, 96, 80, 85, 91, 78, 72, 89]
}

df = pd.DataFrame(data)
print(df.groupby("Department")["Marks"].mean())
print(df.groupby("Department")["Attendance"].mean())
```

```python
# GROUPBY + LAMBDA
# Use groupby("Department") and apply a lambda function
# to calculate the difference between the highest and lowest
# Marks in each department.
# Display:
# Department
# Marks Range
# Hint:
# groupby("Department")["Marks"].apply(...)
# Write your code below.

import pandas as pd
data = {
    "Name": ["Ali", "Meesum", "Shifaan", "Fiza", "Kabir", 
             "Danish", "Kumail", "Rukbaan", "Sharik", "Zoya"],
    "Department": ["IT", "CS", "IT", "CS", "ECE",
                   "ECE", "IT", "CS", "ECE", "IT"],
    "Marks": [78, 88, 65, 95, 72, 84, 90, 76, 69, 82],
    "Attendance": [88, 92, 75, 96, 80, 85, 91, 78, 72, 89]
}

df = pd.DataFrame(data)
print(df.groupby("Department")["Marks"].apply(lambda x:x.max()-x.min()))
```

```python
# FINAL MINI CODING CHALLENGE
# Create a function called department_report(df, *departments, **options).
# Requirements:
# 1. Use *departments to receive one or more department names.
# 2. Use **options to allow the user to specify whether they want
#    "marks", "attendance", or both in the report.
# 3. Use pandas groupby() to calculate department-wise averages.
# 4. Use exception handling for an invalid department or invalid option.
# 5. Return/print a clean department-wise report.
# Example idea:
# department_report(df, "IT", "CS", marks=True)
# Keep the solution simple and use only concepts taught in this exercise.

import pandas as pd
data = {
    "Name": ["Ali", "Meesum", "Shifaan", "Fiza", "Kabir",
             "Danish", "Kumail", "Rukbaan", "Sharik", "Zoya"],
    "Department": ["IT", "CS", "IT", "CS", "ECE",
                   "ECE", "IT", "CS", "ECE", "IT"],
    "Marks": [78, 88, 65, 95, 72, 84, 90, 76, 69, 82],
    "Attendance": [88, 92, 75, 96, 80, 85, 91, 78, 72, 89]
}

df = pd.DataFrame(data)


def department_report(df, *departments, **options):
    #print(departments)
    #print(options)
    dept=df[df["Department"].isin(departments)]
    print(dept[["Department","Name","Marks"]])
   
   


print(df.groupby("Department")["Marks"].mean())
department_report(df,"IT","CS",Marks="true",Attendance="true")
```
