```python
### 1. Check Balance 
#2. Deposit Money 
#3. Withdraw Money 
#4. Exit 
#Enter your choice: 1 
#Current Balance: ₹10000 
#Enter your choice: 2 
#Enter amount: 3000 
#₹3000 deposited successfully. 
#Current Balance: ₹13000 
#Enter your choice: 3 
#Enter amount: 2000 
#₹2000 withdrawn successfully. 
#Current Balance: ₹11000 
#Enter your choice: 4 
#Thank you for using the ATM! 
#Final Balance: ₹11000 '''
def deposit(amount,balance):
    balance+=amount
    print("{} deposited successfully".format(amount))
    print("Current Balance : {}".format(balance))
    return(balance)



def withdraw(balance,amnt):
    balance-=amnt
    print("{} withdrawn successfully".format(amnt))
    print("Current Balance : {}".format(balance))
    return(balance)





print("enter 1 for Current Balance")
print("enter 2 for deposit")
print("enter 3 for withdrawal")
print("enter 4 for checking the final balance")
print("enter exit to leave")

balance=20000
while 1:
    choice=str(input("enter your choice"))
    #choice=str(choice)
    if choice=='1':
        print("current Balance:",balance)
    elif choice=='2':
        amount=int(input("enter amount to be deposited"))
        bal=deposit(amount,balance)
        balance=bal
    elif choice=='3':
        amnt=int(input("Enter amount to be withdrawn"))
        if amnt>balance:
            print("minimum balance")
        else:
            bal=withdraw(balance,amnt)
            balance=bal
    elif choice=='4':
        print("Thankyou for using the ATM")
        print("Final Balance :",bal)
    elif  choice=='exit':
        print("Thanks for using ATM")
        break
    else:
        print("you have entered wrong option")
        #break
            
    
         
    
    
    
    
    
```
