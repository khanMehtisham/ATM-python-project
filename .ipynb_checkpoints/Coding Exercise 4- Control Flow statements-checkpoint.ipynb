## Do these programs using control flow statements

```python
#1. Check if a number is even or odd
#num = 10
num=10
if num%2==0:
    print("number is Even")
else:
    print("number is odd")
```

```python
#Compare two numbers for greater than less than conditions 
#a = 30
#b = 20
a=30
b=20
if a>b:
    print("a is greater than b")
else:
    print("a is lesser than b")
```

```python
#3. Check if a number is positive, negative, or zero
#num = -5
num=-5
if num>=0:
    print("number is positive")
else:
    print("number is negative")
```

```python
#4. Check if a person is eligible to vote (age 18 or older)
#age = 21
age=21
if age>=18:
    print("Eligible to vote")
else:
    print("Not Eligible")

```

```python
#5. Determine the largest of three numbers
#a = 10
#b = 30
#c = 20
a=45
b=40
c=50
if a>b and a>c:
    print("a is largest of three numbers")
elif b>c:
    print("b is largest of three numbers")
else:
    print("c is largest of three numbers")
 
```
